## 1. Objective

The objective of this notebook is to understand the Fetal Health Classification dataset before applying machine learning algorithms.

We will perform Exploratory Data Analysis (EDA) to understand the structure, quality, distribution, and characteristics of the dataset.

## 2. Dataset Description

The dataset contains Cardiotocography (CTG) observations represented through numerical features.

Each record contains measurements describing different characteristics of fetal heart rate and uterine activity.

The dataset contains 21 input features and one target variable named `fetal_health`.

## 3. Prediction Objective

Our objective is to classify each observation into one of three fetal health categories:

- 1: Normal
- 2: Suspect
- 3: Pathological

This is a supervised multiclass classification problem because the dataset contains input features along with their corresponding class labels.

## 8. Target Class Distribution

The target variable `fetal_health` represents three fetal health
categories:

- 1: Normal
- 2: Suspect
- 3: Pathological

We analyze the distribution of these classes to understand
whether the dataset is balanced or imbalanced.

```python
df['fetal_health'].value_counts().sort_index()
```

```python
df['fetal_health'].value_counts(normalize=True).sort_index() * 100
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='fetal_health',
    hue='fetal_health',
    palette='viridis',
    legend=False
)

plt.title('Distribution of Fetal Health Classes')
plt.xlabel('Fetal Health Class')
plt.ylabel('Number of Records')
plt.xticks([0, 1, 2], ['Normal (1)', 'Suspect (2)', 'Pathological (3)'])

plt.show()
```

## 9. Feature Correlation Analysis

Correlation analysis helps us understand the relationships
between numerical features in the dataset.

A correlation value close to +1 indicates a strong positive
relationship, while a value close to -1 indicates a strong
negative relationship. Values near 0 indicate weak linear
relationships.

```python
correlation_matrix = df.corr(numeric_only=True)

plt.figure(figsize=(18, 14))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Fetal Health Dataset')
plt.show()
```

## 10. Feature Distribution Analysis

We visualize the distributions of numerical features to
understand their spread, skewness, and possible unusual values.

This helps us identify differences in feature scales and
patterns that may be relevant during preprocessing.

```python
df.drop(columns=['fetal_health']).hist(
    figsize=(20, 18),
    bins=30,
    edgecolor='black'
)

plt.suptitle('Distribution of Input Features', fontsize=16)
plt.tight_layout()
plt.show()
```

## 11. Outlier Visualization

Boxplots are used to visualize the spread of numerical features
and identify potential outliers.

Outliers identified through boxplots are not automatically
removed. Their impact will be evaluated during preprocessing
and model development.

```python
features = df.drop(columns=['fetal_health']).columns

plt.figure(figsize=(20, 12))

for i, column in enumerate(features, 1):
    plt.subplot(5, 5, i)
    sns.boxplot(y=df[column])
    plt.title(column)

plt.tight_layout()
plt.show()
```
